```python
from fit_integrators import (
    load_trial_data,
    fit_perfect, predict_proba_perfect,
    fit_leaky, predict_proba_leaky,
    fit_confirmation, predict_proba_confirmation,
    fit_normative, predict_proba_normative,
)
from evaluate_models import evaluate_model, build_model_results_df

MODELS = {
    'perfect':      (fit_perfect,      predict_proba_perfect),
    'leaky':        (fit_leaky,        predict_proba_leaky),
    'confirmation': (fit_confirmation, predict_proba_confirmation),
    'normative':    (fit_normative,    predict_proba_normative),
}

def evaluate_all(d_matrix, y):
    """Fit and score every model on (d_matrix, y), in-sample."""
    results = {name: evaluate_model(fit_fn, proba_fn, d_matrix, y, d_matrix, y)
               for name, (fit_fn, proba_fn) in MODELS.items()}
    return build_model_results_df(results)

d_matrix, y, df = load_trial_data('results.csv')

subsets = {
    'combined': df.index.values,
    'H = 0':    df.index[df['H'] == 0.0].values,
    'H = 0.1':  df.index[df['H'] == 0.1].values,
}

model_results = {}
for label, idx in subsets.items():
    model_results[label] = evaluate_all(d_matrix[idx], y[idx])
    print(f'===== {label}  (n = {len(idx)}) =====')
    display(model_results[label])
    print(f"best by AIC: {model_results[label]['aic'].astype(float).idxmin()}   "
          f"best by BIC: {model_results[label]['bic'].astype(float).idxmin()}\n")
```
